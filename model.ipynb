```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import pickle
```

```python
data = pd.read_csv("F:\\Github_Desktop\\UNSUPERVISED-LEARNING-ALGORIDHAM\\mobile_recommendation_system_dataset copy.csv")
print("✅ Dataset Loaded Successfully!")
display(data.head())
```

```python
numeric_features = data.select_dtypes(include=['number']).columns.tolist()
print("Numeric Features:", numeric_features)
```

```python
X = data[numeric_features]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
k = 4   # you can change clusters later if needed
model = KMeans(n_clusters=k, random_state=42)
data['Cluster'] = model.fit_predict(X_scaled)
print("✅ Model Trained Successfully with", k, "clusters")
```

```python
with open("model.pkl", "wb") as file:
    pickle.dump((model, scaler, numeric_features), file)

print("✅ Model saved as model.pkl")

display(data.head())
```
